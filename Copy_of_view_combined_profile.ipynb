```python
# @title Setup - Markdown Decoder

from IPython.display import display, Markdown
import re

def decode_and_display_markdown(encoded_text: str):
  """Decodes mojibake text and displays as Markdown."""
  if not encoded_text or not isinstance(encoded_text, str):
    print("Text is empty or not a string.")
    return

  try:
    # First, try to fix octal escape sequences (like \303\250)
    decoded = encoded_text.encode('latin-1').decode('unicode-escape').encode('latin-1').decode('utf-8')
  except (UnicodeDecodeError, UnicodeEncodeError, AttributeError):
    try:
      # If that fails, try just the mojibake fix
      decoded = encoded_text.encode('latin-1').decode('utf-8')
    except (UnicodeDecodeError, UnicodeEncodeError):
      # If both fail, use original text
      decoded = encoded_text

  # Handle newline characters
  lines = re.split(r'\\n|\n', decoded)
  markdown_content = '\n'.join(lines).strip()

  if markdown_content:
    display(Markdown(markdown_content))
```
